```
import pandas as pd
from scipy import stats
import numpy as np

# 1. 데이터 불러오기 (정확한 경로 설정)
# 왼쪽 파일 탐색기 기준: sample_data 폴더 -> type3 폴더 안에 있는 파일
df = pd.read_csv("sample_data/type3/delivery_time.csv")

# 2. 데이터 전처리 (결측치 제거)
x = df["delivery_min"].dropna()

# 3. 단일표본 t-test 수행
# 귀무가설(H0): 평균 배달시간은 30분 이상이다.
# 대립가설(H1): 평균 배달시간은 30분 미만이다. (alternative='less')
result = stats.ttest_1samp(x, popmean=30, alternative='less')

# 4. 기초 통계량 및 95% 신뢰구간 계산
mean_x = x.mean() # 표본평균
n = len(x)        # 표본 수
se = stats.sem(x) # 표준오차

# t-분포를 이용한 신뢰구간 계산
ci_low, ci_high = stats.t.interval(0.95, df=n-1, loc=mean_x, scale=se)

# 5. 결과 출력
print("-" * 30)
print("   단일표본 t-test 분석 결과   ")
print("-" * 30)
print(f"표본 수: {n}")
print(f"표본평균: {mean_x:.4f}")
print(f"t-통계량(t-statistic): {result.statistic:.4f}")
print(f"p-값(p-value): {result.pvalue:.6f}")
print(f"95% 신뢰구간: ({ci_low:.4f}, {ci_high:.4f})")
print("-" * 30)

# 6. 해석 가이드
if result.pvalue < 0.05:
    print("해석: p-value가 0.05보다 작으므로 귀무가설을 기각합니다.")
    print("결론: 배달시간의 평균은 30분보다 짧다고 할 수 있습니다.")
else:
    print("해석: p-value가 0.05보다 크므로 귀무가설을 기각할 수 없습니다.")
    print("결론: 배달시간의 평균이 30분보다 짧다고 볼 근거가 부족합니다.")
```

```
import pandas as pd
from scipy import stats

# 1. 데이터 불러오기 (정확한 경로로 수정 완료)
# 파일이 sample_data 폴더 안의 type3 폴더에 있으므로 아래와 같이 작성합니다.
df = pd.read_csv("sample_data/type3/before_after_worktime.csv")

# 2. 변수 설정
before = df["before_min"]
after = df["after_min"]

# 3. 대응표본 t-test 수행
# alternative="greater"는 교육 전(before) 시간이 교육 후(after)보다 더 길었는가(즉, 시간이 단축되었는가)를 검정합니다.
diff = before - after
result = stats.ttest_rel(before, after, alternative="greater")

# 4. 결과 출력
print("-" * 40)
print("   대응표본 t-test(Paired t-test) 결과   ")
print("-" * 40)
print(f"평균 차이(before - after): {diff.mean():.4f}분")
print(f"t-통계량(t-statistic): {result.statistic:.4f}")
print(f"p-값(p-value): {result.pvalue:.6f}")
print("-" * 40)

# 5. 결과 해석 가이드
if result.pvalue < 0.05:
    print("해석: p-value가 0.05보다 작으므로 귀무가설을 기각합니다.")
    print("결론: 교육 전후 업무처리시간에 유의미한 차이가 있으며, 교육 후 시간이 단축되었다고 볼 수 있습니다.")
else:
    print("해석: p-value가 0.05보다 크므로 귀무가설을 기각할 수 없습니다.")
    print("결론: 교육 전후 업무처리시간에 유의미한 차이가 있다고 보기 어렵습니다.")
```
